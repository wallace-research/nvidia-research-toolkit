## 1. Download prices

```python
prices = download_price_history(["NVDA", "SPY"], period="5y")
prices.tail()
```

## 2. Normalized performance chart

```python
normalized = prices / prices.iloc[0] * 100
ax = normalized.plot(figsize=(10, 5), title="NVDA vs SPY – Normalized (start = 100)")
ax.set_ylabel("Index")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 3. Return metrics

```python
returns = period_returns(prices)
returns = relative_to_spy(returns)
returns
```

## 4. Fundamental snapshot

```python
info = download_nvda_info()
for k, v in info.items():
    print(f"{k:20}: {v}")
```

## 5. Illustrative DCF

```python
dcf = simple_dcf(
    free_cash_flow=50_000,   # $ millions – illustrative
    growth_rate=0.15,
    terminal_growth=0.03,
    discount_rate=0.10,
    years=5,
)

print(f"Approx. enterprise value: ${dcf['enterprise_value_approx']:,.0f} m")
print("\nYear-by-year:")
for row in dcf["year_by_year"]:
    print(f"  Year {row['year']}: FCF ${row['fcf']:,.0f} m  →  PV ${row['pv']:,.0f} m")
```
